```python
# @title Texto de título predeterminado
%%bash
# 1. Crear carpetas de trabajo y salida en Google Drive
mkdir -p /content/wrapper
mkdir -p "/content/drive/MyDrive/AppleMusic"

# 2. Descargar Wrapper Lite QEMU
cd /content/wrapper
wget -O wrapper-lite-qemu.zip "https://nightly.link/WorldObservationLog/wrapper/workflows/build-lite/lite/wrapper-lite-qemu-linux-x86_64.zip"
unzip -o wrapper-lite-qemu.zip
chmod +x wrapper-lite-qemu

# 3. Descargar amdl
cd /content
curl -L -o amdl https://github.com/zhaarey/apple-music-downloader/releases/latest/download/amdl_linux_amd64
chmod +x amdl

# 4. Generar la configuración apuntando a Google Drive
cat <<EOF > /content/config.yaml
general:
  lite-server: "http://127.0.0.1:12340"
paths:
  alac: "/content/drive/MyDrive/AppleMusic/ALAC"
  atmos: "/content/drive/MyDrive/AppleMusic/Atmos"
  aac: "/content/drive/MyDrive/AppleMusic/AAC"
  mv: "/content/drive/MyDrive/AppleMusic/MV"
EOF
```
